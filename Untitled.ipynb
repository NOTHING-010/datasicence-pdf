```python
#ex 2 b
def clean_str(s, rem_list): 
    s = s.lower() 
    for ch in rem_list: 
      s = s.replace(ch, "") 
    return s 
def word_freq(s): 
   words = s.split() 
   counts = [] 
   for w in words: 
      counts.append(words.count(w)) 
   return dict(zip(words, counts)) 
def show_count(w, freq_dict): 
   w = w.lower() 
   if w in freq_dict: 
     return freq_dict[w] 
   else: 
     return 0 
if __name__ == "__main__": 
  inp = "She sells seashells on the sea shore. The shells she sells are seashells, I'm sure. And if she sells seashells on the sea shore, Then I'm sure she sells seashore shells." 
  rem = [".", ",", "?"] 
  clean = clean_str(inp, rem) 
  freq = word_freq(clean) 
  test = "if" 
  print("the word in the para:",show_count(test, freq))  
```

# PROCESSING TEXT FILES  

```python
def read_process(file_name): 
    words = [] 
    with open(file_name, "rt",encoding="utf-8") as f: 
        words = [word for line in f for word in line.split()] 
     
    # Convert to lowercase
    words = [w.lower() for w in words] 
    
    # Characters to clean
    char_clean = "!;:'\"\\, ./?@#$%^&*_~"
    clean_words = [] 
     
    # Remove unwanted characters
    for word in words: 
        for char in word: 
            if char in char_clean: 
                word = word.replace(char, "") 
        clean_words.append(word) 
     
    return clean_words 
 
def count_freq(words, word_to_count): 
    """Count frequency of a given word in the list"""
    return words.count(word_to_count.lower()) 

if __name__ == "__main__": 
    word_list = read_process(r"D:\DATA SICENCE LAB\dataset\sample.txt") 
    print("THE NO OF WORD :",count_freq(word_list, "the"))
```

```python
import pandas as pd 
import datetime as dt 

def get_hl(df): 
    df.drop(df.columns[-1], axis=1, inplace=True) 
    df["Date"] = pd.to_datetime(df["Date"], format='%d-%B-%Y') 
    df.fillna(0, inplace=True) 
    s = dt.datetime.strptime('2018-03-31', '%Y-%m-%d') 
    e = dt.datetime.strptime('2019-04-01', '%Y-%m-%d') 
    df_fy = df[(df["Date"] > s) & (df["Date"] < e)] 
    hi = df_fy["High"].max() 
    lo = df_fy["Low"].min() 
    return hi, lo, df_fy 

if __name__ == "__main__": 
    df = pd.read_csv(R"D:\DATA SICENCE LAB\dataset\SENSEX_01012018_31122019.csv") 
    hi, lo, df_fy = get_hl(df) 
    df_fy.to_csv("sensex_fy2019-20.csv", index=False) 
    print("SENSEX High & Low in FY2019-20:", hi, "&", lo) 
```

#  handle HTML and EXCEL files.  

```python
import pandas as pd

def get_irnss_data(url, table):
    # Read HTML table from Wikipedia page
    data = pd.read_html(url, match=table)
    df = data[0]

    # Filter out rows where Status contains 'Planned'
    df_sub = df[~df['Status'].str.contains('Planned')].copy()

    # Convert Launch Date to datetime (robust parsing)
    df_sub['Launch Date'] = pd.to_datetime(
        df_sub['Launch Date'], 
        format='%d %B %Y', 
        errors='coerce'   # handles irregular formats safely
    )

    # Sort by Launch Date (latest first)
    df_sub = df_sub.sort_values(by='Launch Date', ascending=False)

    # Convert back to string format for neat Excel output
    df_sub['Launch Date'] = df_sub['Launch Date'].dt.strftime('%d %B %Y')

    return df_sub

if __name__ =="_main_":
    url = "https://en.wikipedia.org/wiki/Indian_Regional_Navigation_Satellite_System"
    table = "IRNSS-1 series satellites"

    # Get cleaned IRNSS data
    result_df = get_irnss_data(url, table)

    # Export to Excel (explicit engine for clarity)
    result_df.to_excel(
        'result.xlsx', 
        sheet_name='IRNSS Launch', 
        index=False, 
        engine='openpyxl'
    )

```

```python
import pandas as pd
import requests

def get_irnss_data(url, table):
    # Pretend to be a browser
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)
    response.raise_for_status()  # raises error if request fails

    # Parse HTML tables
    data = pd.read_html(response.text, match=table)
    df = data[0]

    # Filter out planned launches
    df_sub = df[~df['Status'].str.contains('Planned')].copy()

    # Convert Launch Date safely
    df_sub['Launch Date'] = pd.to_datetime(
        df_sub['Launch Date'], 
        format='%d %B %Y', 
        errors='coerce'
    )

    # Sort and format
    df_sub = df_sub.sort_values(by='Launch Date', ascending=False)
    df_sub['Launch Date'] = df_sub['Launch Date'].dt.strftime('%d %B %Y')

    return df_sub

url = "https://en.wikipedia.org/wiki/Indian_Regional_Navigation_Satellite_System"
table = "IRNSS-1 series satellites"
result_df = get_irnss_data(url, table)
result_df.to_excel('result.xlsx', sheet_name='IRNSS Launch', index=False, engine='openpyxl')
```
